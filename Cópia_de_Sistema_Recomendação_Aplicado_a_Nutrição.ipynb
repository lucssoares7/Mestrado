Calculadora de Macronutrientes

```python
def determinar_grupo_etario(idade):
    """Determina o grupo etário com base na idade do paciente."""
    if idade <= 18:
        return 'pediatrico'
    elif idade >= 60:
        return 'idoso'
    else:
        return 'adulto'

def calcular_necessidades_nutricionais(peso, idade, imc, sexo, fase_tratamento, atividade_fisica, consistencia_alimentar):
    grupo_etario = determinar_grupo_etario(idade)

    # Seção 1: Macronutrientes
    macronutrientes = {}

    # Calorias e Proteínas Ajustadas à Fase de Tratamento
    if fase_tratamento == 'tratamento_ativo':
        # Necessidades aumentadas devido ao estresse metabólico
        calorias = 25 * peso
        proteinas = 1.5 * peso if imc < 30 else 2.0 * peso if imc < 40 else 2.5 * peso
    elif fase_tratamento == 'recuperacao':
        # Necessidades para reconstrução do estado nutricional
        calorias = 25 * peso
        proteinas = 1.2 * peso if imc < 30 else 1.5 * peso if imc >= 30 else 1.8 * peso
    elif fase_tratamento == 'manutencao_saude':
        # Manutenção da saúde com prevenção de recorrência e equilíbrio calórico
        calorias = 22 * peso
        proteinas = 1.0 * peso
    elif fase_tratamento == 'doenca_avancada':
        # Foco em conforto e palatabilidade
        calorias = 20 * peso
        proteinas = 0.8 * peso
    else:
        # Valor padrão caso fase de tratamento não seja especificada corretamente
        calorias = 0
        proteinas = 0

    macronutrientes['calorias'] = calorias
    macronutrientes['proteinas'] = proteinas

    # Gorduras e carboidratos
    gordura_total = (20, 35) if grupo_etario != 'pediatrico' else (30, 40)
    gordura_saturada = "<10% do VET"
    carboidratos = (50, 60)
    macronutrientes['gordura_total'] = gordura_total
    macronutrientes['gordura_saturada'] = gordura_saturada
    macronutrientes['carboidratos'] = carboidratos

    # Hidratação
    hidrico = 30 * peso if grupo_etario == 'adulto' else (25 * peso if grupo_etario == 'idoso' else
              (100 * peso if peso <= 10 else 1000 + 50 * (peso - 10) if peso <= 20 else 1500 + 20 * (peso - 20)))
    macronutrientes['hidrico'] = hidrico

    # Consistência alimentar
    refeicoes_diarias = consistencia_alimentar if consistencia_alimentar else 5
    macronutrientes['refeicoes_diarias'] = refeicoes_diarias
    macronutrientes['calorias_por_refeicao'] = round(calorias / refeicoes_diarias, 2)

    # Seção 2: Recomendações Específicas
    recomendacoes = {}
    recomendacoes['frutas_vegetais'] = "Consumir 3 porções de frutas e 3 porções de legumes/verduras, totalizando pelo menos 400 g/dia."
    recomendacoes['carne_vermelha'] = (
        "Para adultos: limitar a 500 g/semana e evitar carnes processadas. "
        "Para crianças/adolescentes: consumir até 3 porções/semana."
    )
    recomendacoes['alimentos_processados_e_sal'] = "Evitar alimentos processados e limitar o sal a 5 g/dia (2 g de sódio)."
    recomendacoes['alcool'] = (
        "Para adultos, limitar a 1 dose/dia para mulheres e 2 doses/dia para homens. "
        "Crianças, adolescentes e gestantes devem evitar o consumo de bebidas alcoólicas."
    )

    # Seção 3: Micronutrientes e Suplementos
    micronutrientes = {}
    micronutrientes['vitaminas'] = {
        'Vitamina C': "90 mg para homens, 75 mg para mulheres",
        'Vitamina D': "800 UI/dia se exposição solar insuficiente",
        'Zinco': "11 mg para homens, 8 mg para mulheres",
        'Ferro': "8 mg para homens, 18 mg para mulheres (ajustado conforme perda sanguínea)"
    }
    micronutrientes['suplementacao'] = (
        "Suplementação apenas em caso de carências nutricionais diagnosticadas."
    )

    return macronutrientes, recomendacoes, micronutrientes

# Exemplo de uso
peso = 70  # Peso em kg
altura = 1.75  # Altura em metros
idade = 65  # Idade em anos
imc = peso / (altura ** 2)  # Cálculo do IMC
sexo = 'F'  # Sexo do paciente
fase_tratamento = 'recuperacao'  # Fase do tratamento
atividade_fisica = 'Levemente Ativo'  # Nível de atividade física
consistencia_alimentar = 5  # Número de refeições diárias

macronutrientes, recomendacoes, micronutrientes = calcular_necessidades_nutricionais(
    peso, idade, imc, sexo, fase_tratamento, atividade_fisica, consistencia_alimentar
)

# Exibindo as necessidades nutricionais
print("Macronutrientes:")
for key, value in macronutrientes.items():
    print(f"{key.capitalize()}: {value}")

print("\nRecomendações Específicas:")
for key, value in recomendacoes.items():
    print(f"{key.capitalize()}: {value}")

print("\nMicronutrientes e Suplementos:")
for key, value in micronutrientes.items():
    print(f"{key.capitalize()}: {value}")
```

```python
import os
import csv
import numpy as np
from tabulate import tabulate
from itertools import groupby
import pandas as pd

nome_arquivo_saida = "alimentos_possiveis.csv"

class TabelaAlimentos:
    def __init__(self, nome_arquivo):
        self.nome_arquivo = nome_arquivo
        #self.colunas_relevantes = ['ALIMENTO', 'Energia Kcal', 'Proteína', 'carboidrato total', 'Lipídios']
        self.colunas_relevantes = [
            'ID', 'ALIMENTO', 'Energia KJ', 'Energia Kcal', 'Umidade g',
            'carboidrato total', 'carboidrato disponível', 'Proteína',
            'Lipídios', 'Fibra alimentar', 'Álcool', 'Cinzas', 'Colesterol',
            'Ácidos graxos saturados', 'Ácidos graxos monoinsaturados',
            'Ácidos graxos polinsaturados', 'Ácidos graxos trans', 'Cálcio',
            'Ferro', 'Sódio', 'Magnésio', 'Fósforo', 'Potássio', 'Zinco',
            'Cobre', 'Selênio', 'Vitamina A (RE)', 'Vitamina A (RAE)',
            'Vitamina D', 'Alfa-tocoferol (Vitamina E)', 'Tiamina', 'Riboflavina',
            'Niacina', 'Vitamina B6', 'Vitamina B12', 'Vitamina C',
            'Equivalente de folato', 'Sal de adição', 'Açúcar de adição',
            'Tipo', 'Estado'
        ]
        self.colunas_relevantes = [col.upper() for col in self.colunas_relevantes]
        self.dados = self.carregar_dados()
        self.grupos = {
          "Cereais, Pães, Tubérculos e Raízes": [
              "ARROZ", "PÃO", "MASSA", "BATATA", "MANDIOCA", "MILHO", "FARINHA","BISCOITO SAlGADO","CANELONE","CUSCUZ","CEREAL","CEREAIS","PANQUECA","PIZZA","TORRADA","WAFFLE","PASTEL"
              "TRIGO", "AVEIA", "CEVADA", "QUINOA", "BATATA-DOCE", "INHAME", "COUSCOUS", "PIPOCA","TORTA", "FAROFA","LASANHA","MACARRÃO","SEMENTE","TAPIOCA","SANDUICHE","SANDUÍCHE","PÃO DE ALHO","TORTEI","RONDELE","RAVIOLI","PURÊ"
          ],
          "Hortaliças (Legumes e Verduras)": [
            "ALFACE", "TOMATE", "CENOURA", "BRÓCOLIS", "ESPINAFRE", "PEPINO","CEBOLA","COGUMELO","LEGUMES","PIMENTÃO",
              "ABOBRINHA", "COUVE", "COUVE-FLOR", "RÚCULA", "NABO", "BETERRABA", "REPOLHO", "CHUCHU", "ALHO-PORÓ", "ABÓBORA","VEGETAIS","SALADA","BETERRABA","BERINJELA","ACELGA","AGRIÃO","ALMEIRÃO","QUIABO"
          ],
          "Frutas": [
              "MAÇÃ", "BANANA", "LARANJA", "MORANGO", "UVA", "PERA", "MANGA",
              "MELANCIA", "MARACUJÁ", "ABACAXI", "KIWI", "PÊSSEGO", "GOIABA","TANGERINA"
              "AMEIXA", "FIGO", "FRAMBOESA", "CEREJA", "LIMÃO","ABACATE","ACEROLA","CAJU","JACA","CAJA","MAMÃO","ROMÃ"
          ],
          "Leite e Derivados": [
              "LEITE", "QUEIJO", "IOGURTE", "MANTEIGA", "REQUEIJÃO", "CREME DE LEITE",
              "RICOTA", "LEITE CONDENSADO", "LEITE EM PÓ", "COALHADA","MUÇARELA"
          ],
          "Carnes e Ovos": [
              "CARNE", "FRANGO", "OVO","PORCO", "BOI", "FRANGO", "PEITO DE PERU", "HAMBÚRGUER", "TOFU","OVOS",
              "SOJA", "ERVILHA", "BACON","STROGONOFF", "STEAK","BACON","ROUPA VELHA","CHARQUE","APRESUNTADO","BIFE","CHURRASQUINHO","DOBRADINHA","OMELETE","PAPA DE FÍGADO","PAPA DE PEITO","FÍGADO"
          ],
          "Leguminosas":[
               "FEIJÃO", "LENTILHA", "GRÃO-DE-BICO","VAGEM","ERVILHA","SOHA"
          ],
          "Óleos e Gorduras": [
              "ÓLEO", "AZEITE", "MANTEIGA", "MARGARINA", "NOZES",
              "CASTANHA", "AMÊNDOAS", "SEMENTE DE GIRASSOL", "SEMENTE DE LINHAÇA",
              "SEMENTE DE ABÓBORA", "BANHA", "ÓLEO DE COCO","CREME","MAIONESE","AMENDOIM","AMÊNDOAS","SEMENTE DE GIRASSOL","SEMENTE DE LINHAÇA","SEMENTE DE ABÓBORA","BANHA","ÓLEO DE COCO"
          ],
          "Açúcares e Doces": [
              "AÇÚCAR", "MEL", "DOCE", "CHOCOLATE", "SORVETE","BOMBOM",
              "BALAS", "BISCOITO RECHEADO", "TORTA", "BOLO", "PUDIM", "GELATINA","BISCOITO DOCE","MARIA MOLE"
              "GOMAS DE MASCAR", "CREME DE AVELÃ", "GELÉIA","ADOÇANTE","BALA","XAROPE","SUNDAE","CARAMELO","ACHOCOLATADO","BRIGADEIRO","CANUDINHO","CHURROS","BOMBA","DOCE","ARROZ DOCE"
          ],
          "Peixes e Frutos do Mar": [
              "PEIXE", "CAMARÃO", "MARISCO", "SIRI", "LULA", "POLVO",
              "SALMÃO", "ATUM", "BACALHAU", "TRUTA", "TILÁPIA", "SARDINHA",
              "CAVALA", "ANCHOVA", "OSTRA", "MEXILHÃO","FRUTOS DO MAR"
          ],
          "Bebidas":[
              "BEBIDA","DRINK","BEBIDA ALCOÓLICA","REFRIGERANTE","CAFÉ","ALCOÓLICA","SUCO","ÁGUA","ACHOCOLATADO"
          ]
        }

    def carregar_dados(self):
        try:
            with open(self.nome_arquivo, newline='', mode='r', encoding='ISO-8859-1') as arquivo_csv:
                leitor_csv = csv.DictReader(arquivo_csv, delimiter=',')
                dados_upper = []

                for linha in leitor_csv:
                    linha_processada = {}
                    for coluna, valor in linha.items():
                        if coluna == 'ALIMENTO' or coluna == 'TIPO':
                            linha_processada[coluna] = valor.upper() if valor else valor
                        else:
                            linha_processada[coluna] = valor

                    dados_upper.append(linha_processada)

                # Ordenar dados pela coluna 'ALIMENTO'
                dados_ordenados = sorted(dados_upper, key=lambda x: x.get('ALIMENTO', '').upper())

            return dados_ordenados

        except FileNotFoundError:
            print(f"Erro: Arquivo {self.nome_arquivo} não encontrado.")
        except Exception as e:
            print(f"Ocorreu um erro: {e}")


    def definir_estado_fisico(self):
        # Dicionário de palavras-chave para estados físicos
        estados = {
            "SÓLIDO": ["PÃO", "QUEIJO", "CARNE", "FRANGO", "BATATA", "ARROZ", "MASSA", "FRUTA", "LEGUME"],
            "LÍQUIDO": ["LEITE", "ÁGUA", "SUCO", "REFRIGERANTE", "BEBIDA", "CAFÉ" "CALDO", "SOPA", "IOGURTE", "VINHO"]
        }

        for alimento in self.dados:
            nome = alimento.get('ALIMENTO', '').upper().strip()  # Garante que o nome esteja em uppercase e sem espaços extras
            estado_definido = False

            # Verifica se o nome do alimento contém uma palavra-chave associada a um estado físico
            for estado, palavras_chave in estados.items():
                if any(palavra in nome for palavra in palavras_chave):
                    alimento['ESTADO'] = estado  # Atribui o estado físico correspondente
                    estado_definido = True
                    break

            if not estado_definido:
                alimento['ESTADO'] = "DESCONHECIDO"  # Caso o estado físico não seja identificado

    def exibir_todos_os_dados(self, max=100):
        for linha in self.dados[:max]:
            print(linha)

    def getnomesAlimentos(self):
        return [alimento.get('ALIMENTO', '').strip() for alimento in self.dados]

    def getlistaAlimentos(self):
        return self.dados

    def adicionar_alimento(self, novo_alimento):
        if novo_alimento:
            self.dados.append({coluna: novo_alimento[coluna] for coluna in self.colunas_relevantes})
            self.salvar_dados(self.gerar_nome_arquivo())

    def preencher(self, lista_alimentos):
        #self.colunas_relevantes = ['ALIMENTO', 'Energia Kcal', 'Proteína', 'carboidrato total', 'Lipídios']
        '''
        self.colunas_relevantes = [
            'ID', 'ALIMENTO', 'Energia KJ', 'Energia Kcal', 'Umidade g',
            'carboidrato total', 'carboidrato disponível', 'Proteína',
            'Lipídios', 'Fibra alimentar', 'Álcool', 'Cinzas', 'Colesterol',
            'Ácidos graxos saturados', 'Ácidos graxos monoinsaturados',
            'Ácidos graxos polinsaturados', 'Ácidos graxos trans', 'Cálcio',
            'Ferro', 'Sódio', 'Magnésio', 'Fósforo', 'Potássio', 'Zinco',
            'Cobre', 'Selênio', 'Vitamina A (RE)', 'Vitamina A (RAE)',
            'Vitamina D', 'Alfa-tocoferol (Vitamina E)', 'Tiamina', 'Riboflavina',
            'Niacina', 'Vitamina B6', 'Vitamina B12', 'Vitamina C',
            'Equivalente de folato', 'Sal de adição', 'Açúcar de adição',
            'Tipo', 'Estado'
        ]
        self.colunas_relevantes = [col.upper() for col in self.colunas_relevantes]
        '''
        self.dados = lista_alimentos
        self.salvar_dados(self.gerar_nome_arquivo())

    def removerAlimento(self, lista_alimentos_restritos):

        alimentos_atualizados = []
        for alimento in self.dados:
            if alimento not in lista_alimentos_restritos:
                alimentos_atualizados.append(alimento)
        self.dados = alimentos_atualizados
        nome_arquivo_saida = self.gerar_nome_arquivo()
        self.salvar_dados(nome_arquivo_saida)
        return nome_arquivo_saida

    def salvar_dados(self, nome_arquivo_saida=None):
        with open(nome_arquivo_saida, 'w', encoding='ISO-8859-1', newline='') as arquivo_csv:
            escritor_csv = csv.DictWriter(arquivo_csv, fieldnames=self.colunas_relevantes)
            escritor_csv.writeheader()
            escritor_csv.writerows(self.dados)

    def gerar_nome_arquivo(self):
        nome_base, extensao = os.path.splitext(self.nome_arquivo)
        timestamp = datetime.now().strftime("%Y%m%d-%H%M%S")
        novo_nome = f"{nome_base}_atualizado_{timestamp}{extensao}"
        return novo_nome

    def buscar_kcal(self, ID, unidades):
        unidades = round(unidades, 3)
        for alimento in self.dados:
            if alimento['ID'] == ID:
                nome = alimento['ALIMENTO']
                kcal_por_unidade = float(alimento['ENERGIA KCAL'].replace(',', '.'))
                kcal_total = round(kcal_por_unidade * unidades)
                return nome[:15], kcal_por_unidade, unidades, kcal_total
        return None

    def gerar_tabela_latex(self):
        dados_formatados = [{coluna: alimento[coluna] for coluna in self.colunas_relevantes} for alimento in self.dados]
        return tabulate(dados_formatados, headers="keys", tablefmt="latex_raw")

    def agruparTipo(self):
        dados_ordenados_por_tipo = sorted(self.dados, key=lambda x: x['TIPO'])
        grupos_tipos = {}
        for tipo, alimentos_do_tipo in groupby(dados_ordenados_por_tipo, key=lambda x: x['TIPO']):
            grupos_tipos[tipo] = list(alimentos_do_tipo)
        return grupos_tipos

    def gerar_tabelas_latex_por_tipo(self):
        grupos_por_tipo = self.agruparTipo()
        tabelas_latex = {}
        for tipo, alimentos in grupos_por_tipo.items():
            dados_formatados = [{coluna: alimento[coluna] for coluna in self.colunas_relevantes} for alimento in alimentos]
            tabelas_latex[tipo] = tabulate(dados_formatados, headers="keys", tablefmt="latex_raw")
        return tabelas_latex

    def dividir_valores_por_10(self):
        colunas_numericas = [
            'Energia KJ', 'Energia Kcal', 'Umidade g',
            'carboidrato total', 'carboidrato disponível', 'Proteína',
            'Lipídios', 'Fibra alimentar', 'Álcool', 'Cinzas', 'Colesterol',
            'Ácidos graxos saturados', 'Ácidos graxos monoinsaturados',
            'Ácidos graxos polinsaturados', 'Ácidos graxos trans', 'Cálcio',
            'Ferro', 'Sódio', 'Magnésio', 'Fósforo', 'Potássio', 'Zinco',
            'Cobre', 'Selênio', 'Vitamina A (RE)', 'Vitamina A (RAE)',
            'Vitamina D', 'Alfa-tocoferol (Vitamina E)', 'Tiamina', 'Riboflavina',
            'Niacina', 'Vitamina B6', 'Vitamina B12', 'Vitamina C',
            'Equivalente de folato', 'Sal de adição', 'Açúcar de adição'
        ]

        colunas_numericas = [col.upper() for col in colunas_numericas]
        for alimento in self.dados:
            for coluna in colunas_numericas:
                if coluna in alimento:
                    valor = alimento[coluna]
                    if valor in ['tr', 'NA', '', None, 'NaN']:
                        alimento[coluna] = '0'  # Substitui por '0'
                    else:
                        try:
                            # Substitui vírgula por ponto, converte para float, divide por 10 e arredonda para 3 casas
                            alimento[coluna] = str(round(float(valor.replace(',', '.')) / 10, 3))
                        except ValueError:
                            # Tratar valores não numéricos (se houver) que não foram capturados anteriormente
                            pass

    def classificar_alimentos(self):
        """
        Classifica os alimentos com base em palavras-chave, priorizando `nome1` sobre `nome2`.
        """
        stopwords = {"DE", "EM", "NO", "NA", "DA", "DO", "COM", "PARA", "A", "O", "E"}
        self.grupos = {chave.upper(): [item.upper() for item in valores] for chave, valores in self.grupos.items()}

        for alimento in self.dados:
            nome = alimento.get('ALIMENTO', '').upper().strip()
            nomes = [palavra for palavra in nome.split() if palavra not in stopwords]
            nome1 = nomes[0] if nomes else ''
            if nome1.endswith(','):
                nome1 = nome1[:-1]

            nome2 = nomes[1] if len(nomes) > 1 else ''
            if nome2.endswith(','):
                nome2 = nome2[:-1]

            alimento['TIPO'] = "OUTROS"  # Inicializa como "OUTROS"
            grupo_encontrado = ''

            # Verifica `nome1` nos grupos
            for grupo in self.grupos:
                if nome1 in self.grupos[grupo]:
                    grupo_encontrado = grupo
                    alimento['TIPO'] = grupo_encontrado
                    break

            # Se `nome1` não encontrou, verifica `nome2`
            if not grupo_encontrado and nome2:
                for grupo, palavras_chave in self.grupos.items():
                    if nome2 in palavras_chave:
                        grupo_encontrado = grupo
                        alimento['TIPO'] = grupo_encontrado
                        break

            if not grupo_encontrado:
                alimento['TIPO'] = "OUTROS"
            # Depuração
            #print(nome1, '-', nome2, '-', grupo_encontrado)


    def gerar_matriz_nutricional(self):
        matriz_nutricional = [
            [alimento['ALIMENTO'],
             float(alimento['Energia Kcal'].replace(',', '.')),
             float(alimento['Proteína'].replace(',', '.')),
             float(alimento['carboidrato total'].replace(',', '.')),
             float(alimento['Lipídios'].replace(',', '.')),
             float(alimento['Fibra alimentar'].replace(',', '.'))]
            for alimento in self.dados
        ]
        return np.array(matriz_nutricional)

    def copy(self):
        return TabelaAlimentos(self.nome_arquivo)
    def buscarAlimentosRestritos(self, lista_alimentos_restritos):
        # Verifica se a lista de alimentos restritos é válida
        if not isinstance(lista_alimentos_restritos, list):
            raise ValueError("A lista de alimentos restritos deve ser do tipo lista.")

        # Cria uma lista para armazenar os alimentos restritos
        alimentos_restritos_encontrados = []

        # Para cada alimento nos dados, verifica se está na lista de restrições
        for alimento in self.dados:
            for restricao in lista_alimentos_restritos:
                # Verifica se a restrição contém um grupo de alimentos
                if ':' in restricao:
                    grupo_alimentos = restricao.split(':', 1)[1]
                    # Se o alimento estiver no grupo, adiciona à lista de restritos
                    if alimento in map(str.strip, grupo_alimentos.split(',')):
                        alimentos_restritos_encontrados.append(alimento)
                        break  # Para evitar duplicação, sai do loop
                else:
                    # Se a restrição é um alimento simples, verifica diretamente
                    if alimento == restricao.strip():
                        alimentos_restritos_encontrados.append(alimento)
                        break  # Para evitar duplicação, sai do loop

        return alimentos_restritos_encontrados

    def getGrupoAlimentos(self, grupo):
        return [alimento for alimento in self.dados if alimento.get('TIPO', '').upper() == grupo.upper()]
```

```python
from datetime import datetime

def calcular_idade(data_nascimento):
    hoje = datetime.now().date()
    return hoje.year - data_nascimento.year - ((hoje.month, hoje.day) < (data_nascimento.month, data_nascimento.day))

def calcularPeso(sexo, data_nascimento, altura):
    idade = calcular_idade(data_nascimento)

    if sexo == 'm':
        probabilidades = [0.017, 0.381, 0.385, 0.218]
    else:
        probabilidades = [0.015, 0.381, 0.331, 0.295]

    categoria = random.choices(
        ['deficit', 'normal', 'excesso', 'obesidade'],
        probabilidades
    )[0]

    # Atribuir um IMC com base na idade e na categoria
    if categoria == 'deficit':
        imc = random.uniform(16.0, 18.4)
    elif categoria == 'normal':
        imc = random.uniform(18.5, 24.9)
    elif categoria == 'excesso':
        imc = random.uniform(25.0, 29.9)
    else:
        imc = random.uniform(30.0, 40.0)


    if idade > 60:
        imc *= 0.95
    elif idade < 25:
        imc *= 1.05

    peso = imc * (altura ** 2)
    return round(peso, 2), categoria
```

```python
import random

def definirAltura(sexo, idade):
    # Definir alturas médias em metros para diferentes faixas etárias e sexos
    if sexo == 'M':
        if idade <= 18:
            # Média para adolescentes e jovens
            altura_media = 1.75
            variacao = 0.10
        elif idade <= 60:
            # Média para adultos homens
            altura_media = 1.75
            variacao = 0.07
        else:
            # Média para homens idosos (ligeira redução)
            altura_media = 1.72
            variacao = 0.05
    else:
        if idade <= 18:
            # Média para adolescentes e jovens mulheres
            altura_media = 1.62
            variacao = 0.10
        elif idade <= 60:
            # Média para adultas mulheres
            altura_media = 1.62
            variacao = 0.07
        else:
            # Média para idosas (ligeira redução)
            altura_media = 1.58
            variacao = 0.05

    # Calcular altura com uma leve variação ao redor da média
    altura = round(random.uniform(altura_media - variacao, altura_media + variacao), 2)
    return altura
```

```python
def escolher_alimentos_por_rank(df, limite_calorias, alimentos_por_refeicao, tipo_refeicao, alimentos_utilizados=None):
    # Inicialização das variáveis
    refeicao = []
    calorias_restantes = limite_calorias
    alimentos_adicionados_nome = set()  # Para evitar duplicação por nome principal

    # Se nenhum conjunto de alimentos utilizados for fornecido, cria um conjunto vazio
    if alimentos_utilizados is None:
        alimentos_utilizados = set()

    # Filtra o DataFrame para o tipo de refeição
    df_filtrado = df[df['TIPO_REFEIÇÃO'].str.contains(tipo_refeicao.strip(), case=False, na=False)]

    # Se o filtro for vazio, usa o DataFrame completo
    if df_filtrado.empty:
        df_filtrado = df

    # Filtrar alimentos não utilizados previamente
    df_filtrado = df_filtrado[
        ~df_filtrado['ALIMENTO'].str.split(',').str[0].str.strip().isin(alimentos_utilizados)
    ]

    # Se todos os alimentos foram usados, volta ao conjunto original
    if df_filtrado.empty:
        df_filtrado = df

    # Ordena o DataFrame por uma métrica de relevância (por exemplo, densidade calórica)
    df_filtrado = df_filtrado.sort_values(by='SCORE', ascending=False)

    # Iterar pelos alimentos do DataFrame filtrado
    for _, alimento in df_filtrado.iterrows():
        if len(refeicao) >= alimentos_por_refeicao:
            break  # Limita a quantidade de alimentos na refeição

        # Extrair apenas a primeira palavra do nome do alimento
        nome_alimento = alimento['ALIMENTO'].upper()
        primeira_palavra = nome_alimento.split()[0].strip()
        if primeira_palavra.endswith(","):
            primeira_palavra = primeira_palavra[:-1]

        # Verifica se a primeira palavra já foi adicionada para evitar duplicatas
        if primeira_palavra in alimentos_adicionados_nome:
            continue

        # Verifica se o alimento possui calorias válidas
        energia_kcal = alimento['ENERGIA KCAL']
        if energia_kcal > 0:
            # Ajusta a quantidade considerando as calorias restantes
            quantidade = min(
                max(1, calorias_restantes // energia_kcal),  # Garante pelo menos 1
                5  # Limita a quantidade máxima a 5
            )
        else:
            continue

        if quantidade > 0:
            calorias_ajustadas = energia_kcal * quantidade

            # Adiciona o alimento com a quantidade ajustada
            refeicao.append({
                'ALIMENTO': alimento['ALIMENTO'],
                'Quantidade': quantidade,
                'ENERGIA KCAL': calorias_ajustadas
            })

            # Atualiza calorias restantes e registra os nomes adicionados
            calorias_restantes -= calorias_ajustadas
            alimentos_adicionados_nome.add(primeira_palavra)

    return refeicao, alimentos_adicionados_nome
```

```python
class Refeicao:
    def __init__(self, nome, limite_calorias, alimentos_por_refeicao, dia_semana):
        if not isinstance(limite_calorias, (int, float)) or limite_calorias <= 0:
            raise ValueError("Limite de calorias deve ser um número positivo")
        if not isinstance(alimentos_por_refeicao, int) or alimentos_por_refeicao <= 0:
            raise ValueError("Alimentos por refeição deve ser um número inteiro positivo")

        self.nome = str(nome)
        self.limite_calorias = float(limite_calorias)
        self.alimentos = []
        self.calorias_totais = 0
        self.alimentos_por_refeicao = alimentos_por_refeicao
        self.dia_semana = str(dia_semana)

    def pode_adicionar_alimento(self, calorias_alimento):

        return (self.calorias_totais + calorias_alimento <= self.limite_calorias and
                len(self.alimentos) < self.alimentos_por_refeicao)

    def adicionar_alimento(self, alimento, quantidade=1):

        if not isinstance(quantidade, (int, float)) or quantidade <= 0:
            raise ValueError("Quantidade deve ser um número positivo")

        if 'ENERGIA KCAL' not in alimento or 'ALIMENTO' not in alimento:
            raise ValueError("Alimento deve conter as chaves 'ENERGIA KCAL' e 'ALIMENTO'")

        calorias_ajustadas = alimento['ENERGIA KCAL'] * quantidade

        if self.pode_adicionar_alimento(calorias_ajustadas):
            self.alimentos.append({
                'ALIMENTO': str(alimento['ALIMENTO']),
                'Quantidade': float(quantidade),
                'ENERGIA KCAL': float(calorias_ajustadas)
            })
            self.calorias_totais += calorias_ajustadas
            return True
        return False

    def adicionar_alimento2(self, alimento):

        if 'ENERGIA KCAL' not in alimento or 'ALIMENTO' not in alimento:
            raise ValueError("Alimento deve conter as chaves 'ENERGIA KCAL' e 'ALIMENTO'")

        if self.pode_adicionar_alimento(alimento['ENERGIA KCAL']):
            alimento_copy = alimento.copy()  # Faz uma cópia para não modificar o original
            self.alimentos.append(alimento_copy)
            self.calorias_totais += alimento['ENERGIA KCAL']
            return True
        return False

    def get_calorias_restantes(self):

        return self.limite_calorias - self.calorias_totais

    def get_espacos_disponiveis(self):

        return self.alimentos_por_refeicao - len(self.alimentos)

    def listar_alimentos(self):

        return self.alimentos.copy()  # Retorna uma cópia para evitar modificações externas

    def __str__(self):

        alimentos_info = "\n".join([
            f"{a['Quantidade']}x {a['ALIMENTO']} ({round(a['ENERGIA KCAL'], 2)} kcal)"
            for a in self.alimentos if a['ENERGIA KCAL'] > 0
        ])

        return (
            f"Dia: {self.dia_semana}\n"
            f"Refeição: {self.nome}\n"
            f"Calorias Totais: {round(self.calorias_totais, 2)}/{self.limite_calorias} kcal\n"
            f"Alimentos:\n{alimentos_info}"
        )

    def __repr__(self):
        return (f"Refeicao(nome='{self.nome}', "
                f"limite_calorias={self.limite_calorias}, "
                f"alimentos_por_refeicao={self.alimentos_por_refeicao}, "
                f"dia_semana='{self.dia_semana}')")
```

```python
def gerar_dieta_semanal(paciente, df, alimentos_por_refeicao=None):
    # Se não definido, gerar número aleatório de alimentos por refeição
    if alimentos_por_refeicao is None:
        alimentos_por_refeicao = random.randint(3, 5)

    dieta_semanal = {
        'dias': {},
        'resumo_nutricional': {
            'calorias_total_semana': 0,
            'calorias_media_dia': 0,
            'variedade_alimentos': set()
        }
    }
    # Conjunto para rastrear alimentos usados na semana
    alimentos_utilizados_na_semana = set()

    # Parâmetros de calorias
    calorias_refeicao = paciente.getCalorias_refeicao()

    for dia in dias_semana:
        # Inicialização do dia
        dieta_semanal['dias'][dia] = {
            'refeicoes': [],
            'calorias_total_dia': 0,
            'alimentos_utilizados': set()
        }

        # Conjunto de alimentos já utilizados neste dia
        alimentos_utilizados_dia = set()

        for ref in Tipos_refeicao[:paciente.getConsistencia_alimentar_aceita()]:
            # Combina alimentos utilizados no dia e na semana
            alimentos_utilizados = alimentos_utilizados_dia.union(alimentos_utilizados_na_semana)

            # Filtrar alimentos disponíveis
            df_disponivel = df[~df['ALIMENTO'].str.split(",").str[0].str.strip().isin(alimentos_utilizados)]

            # Escolher os melhores alimentos
            melhores_alimentos, _ = escolher_alimentos_por_rank(
                df_disponivel,
                calorias_refeicao,
                alimentos_por_refeicao,
                ref
            )
            # Criar refeição
            refeicao = Refeicao(ref, calorias_refeicao, alimentos_por_refeicao, dia)

            # Adicionar alimentos à refeição
            for alimento in melhores_alimentos:
                refeicao.adicionar_alimento2(alimento)

            # Atualizar conjuntos de alimentos
            novos_alimentos = {
                alimento['ALIMENTO'].split(',')[0].strip() for alimento in melhores_alimentos
            }
            alimentos_utilizados_dia.update(novos_alimentos)
            alimentos_utilizados_na_semana.update(novos_alimentos)

            # Adicionar à dieta do dia
            dieta_semanal['dias'][dia]['refeicoes'].append(refeicao)
            dieta_semanal['dias'][dia]['calorias_total_dia'] += refeicao.calorias_totais
            dieta_semanal['dias'][dia]['alimentos_utilizados'].update(novos_alimentos)

        # Atualizar resumo nutricional
        dieta_semanal['resumo_nutricional']['calorias_total_semana'] += dieta_semanal['dias'][dia]['calorias_total_dia']
        dieta_semanal['resumo_nutricional']['variedade_alimentos'].update(
            dieta_semanal['dias'][dia]['alimentos_utilizados']
        )

    # Calcular média de calorias
    dieta_semanal['resumo_nutricional']['calorias_media_dia'] = (
        dieta_semanal['resumo_nutricional']['calorias_total_semana'] / len(dias_semana)
    )

    return dieta_semanal
```

```python
def obter_nome_alimento(alimento):
    """
    Extrai um nome formatado para o alimento

    Args:
        alimento (dict): Dicionário com informações do alimento

    Returns:
        str: Nome formatado do alimento
    """
    try:
        # Split do nome pelo vírgula
        partes = alimento['ALIMENTO'].split(',')

        # Captura primeira palavra de cada parte
        nome_primeira_parte = partes[0].strip()

        # Tratar caso haja mais de uma parte
        if len(partes) > 1:
            nome_segunda_parte = partes[1].strip().split()[0].capitalize()
            nome_alimento = f"{nome_primeira_parte} {nome_segunda_parte}"
        else:
            nome_alimento = nome_primeira_parte

        # Remove parênteses e 'C/'
        nome_alimento = nome_alimento.replace('(', '').replace('C/', '')
        nome_alimento = nome_alimento.upper()
        return nome_alimento.strip()

    except (IndexError, KeyError):
        # Tratamento de erro caso a estrutura do alimento seja diferente
        return "Alimento Desconhecido"


def gerar_tabelas_latex_dieta_semanal(dieta_semanal):
    """
    Gera tabelas LaTeX para cada dia da dieta semanal

    Args:
        dieta_semanal (dict): Dicionário com informações da dieta semanal

    Returns:
        str: Código LaTeX com tabelas dos dias
    """
    # String para armazenar o código LaTeX
    latex_tabelas = ""

    # Processar cada dia da semana
    for dia, info_dia in dieta_semanal['dias'].items():
        # Início da tabela para o dia
        latex_tabelas += fr"""
\begin{{table}}[H]
\caption{{{dia}}}
\centering
\begin{{tabular}}{{llrr}}
\toprule
Refeição & Alimento & Quantidade & Calorias (kcal) \\
\midrule
"""

        # Adicionar linhas de alimentos para cada refeição
        for refeicao in info_dia['refeicoes']:
            for alimento in refeicao.alimentos:
                latex_tabelas += (
                    f"{refeicao.nome} & "
                    f"{obter_nome_alimento(alimento)} & "
                    f"{alimento['Quantidade']:.0f} & "
                    f"{alimento['ENERGIA KCAL']:.0f} \\\\\n"
                )

        # Linha de total de calorias do dia
        latex_tabelas += fr"""
\midrule
\multicolumn{{3}}{{l}}{{Total de Calorias do Dia}} & {info_dia['calorias_total_dia']:.0f} \\
\bottomrule
\end{{tabular}}
\end{{table}}
"""

    return latex_tabelas

tabela_latex = gerar_tabelas_latex_dieta_semanal(dieta_semanal)
print(tabela_latex)
```

```python
def selecionar_melhor_solucao(res, df, limite_calorias_min, limite_calorias_max, alimentos_por_refeicao):

    try:
        # Estratégias de seleção
        def estrategia_proteina(solucao):
            return np.argsort(solucao * df['PROTEÍNA'].values)[::-1]

        def estrategia_nota_nutricional(solucao):
            return np.argsort(solucao * df['NOTA_NUTRICIONAL'].values)[::-1]

        def estrategia_aleatoria(solucao):
            return np.random.permutation(len(solucao))

        estrategias = [
            estrategia_proteina,
            estrategia_nota_nutricional,
            estrategia_aleatoria
        ]

        # Iterar por soluções (primeira solução do resultado)
        for solucao in res.X[:1]:  # Usar apenas a primeira solução
            for estrategia in estrategias:
                try:
                    # Aplicar estratégia de seleção
                    indices = estrategia(solucao)

                    # Selecionar alimentos
                    alimentos_selecionados = []
                    for i in indices:
                        quantidade = round(solucao[i])

                        # Pular se quantidade for zero
                        if quantidade == 0:
                            continue

                        # Calcular calorias e verificar limites
                        calorias = df.iloc[i]['ENERGIA KCAL'] * quantidade

                        # Adicionar alimento
                        alimento = {
                            'ALIMENTO': df.iloc[i]['ALIMENTO'],
                            'Quantidade': quantidade,
                            'Calorias': calorias,
                            'Proteína': df.iloc[i]['PROTEÍNA'] * quantidade
                        }
                        alimentos_selecionados.append(alimento)

                        # Parar quando atingir o número desejado
                        if len(alimentos_selecionados) == alimentos_por_refeicao:
                            break

                    # Verificar se encontramos alimentos suficientes
                    if len(alimentos_selecionados) > 0:
                        return alimentos_selecionados

                except Exception as e:
                    print(f"Falha em estratégia de seleção: {e}")

        # Se todas as estratégias falharem
        return []

    except Exception as e:
        print(f"Erro na seleção de melhor solução: {e}")
        return []
```

```python
def selecionar_alimentos_nsga2_avancado(
    df,
    limite_calorias_min,
    limite_calorias_max,
    alimentos_por_refeicao,
    alimentos_utilizados=None,
    verbose=False
):
    """
    Função para selecionar alimentos usando NSGA-II com tratamento de erros aprimorado
    """
    # Cópia segura do DataFrame
    df = df.copy()

    # Garantir existência das colunas necessárias
    colunas_necessarias = [
        'ALIMENTO',
        'ENERGIA KCAL',
        'PROTEÍNA',
        'CARBOIDRATO TOTAL',
        'LIPÍDIOS',
        'FIBRA ALIMENTAR'
    ]

    # Preencher colunas faltantes com zero
    for col in colunas_necessarias:
        if col not in df.columns:
            df[col] = 0.0

    # Filtrar alimentos já utilizados
    if alimentos_utilizados is not None and len(alimentos_utilizados) > 0:
        df = df[~df['ALIMENTO'].isin(alimentos_utilizados)]

    # Verificar se ainda há alimentos disponíveis
    if df.empty:
        if verbose:
            print("Nenhum alimento disponível após filtro.")
        return []

    # Função de normalização segura
    def normalizar_coluna(serie):
        min_val = serie.min()
        max_val = serie.max()

        # Evitar divisão por zero
        if min_val == max_val:
            return pd.Series(1, index=serie.index)

        return (serie - min_val) / (max_val - min_val)

    # Criar nota nutricional
    try:
        df['NOTA_NUTRICIONAL'] = (
            normalizar_coluna(df['PROTEÍNA']) * 0.4 +
            normalizar_coluna(df['FIBRA ALIMENTAR']) * 0.3 +
            (1 - normalizar_coluna(df['ENERGIA KCAL'])) * 0.3
        )
    except Exception as e:
        print(f"Erro ao calcular nota nutricional: {e}")
        return []

    # Estratégia de fallback
    def selecao_fallback():
        # Ordenar por critérios nutricionais
        candidatos = df.sort_values(
            by=['PROTEÍNA', 'NOTA_NUTRICIONAL', 'ENERGIA KCAL'],
            ascending=[False, False, True]
        )

        # Filtrar por limite calórico
        candidatos_filtrados = candidatos[
            (candidatos['ENERGIA KCAL'] >= limite_calorias_min / alimentos_por_refeicao) &
            (candidatos['ENERGIA KCAL'] <= limite_calorias_max / alimentos_por_refeicao)
        ]

        # Se filtro falhar, usar todos os candidatos
        if candidatos_filtrados.empty:
            candidatos_filtrados = candidatos

        # Selecionar top alimentos
        top_alimentos = candidatos_filtrados.head(alimentos_por_refeicao)

        return [
            {
                'ALIMENTO': row['ALIMENTO'],
                'Quantidade': 1,
                'Calorias': row['ENERGIA KCAL'],
                'Proteína': row['PROTEÍNA']
            }
            for _, row in top_alimentos.iterrows()
        ]

    # Verificações de viabilidade
    if len(df) < alimentos_por_refeicao:
        if verbose:
            print(f"Poucos alimentos disponíveis. Total: {len(df)}")
        return selecao_fallback()

    # Implementação do NSGA-II (pymoo)
    try:
        from pymoo.core.problem import Problem
        from pymoo.algorithms.moo.nsga2 import NSGA2
        from pymoo.optimize import minimize

        class ProblemaAlimentosAvancado(Problem):
            def __init__(self, df, limite_calorias_min, limite_calorias_max):
                super().__init__(
                    n_var=len(df),
                    n_obj=4,
                    n_constr=2,
                    xl=0,
                    xu=3
                )
                self.df = df.reset_index(drop=True)
                self.calorias = df['ENERGIA KCAL'].values
                self.proteina = df['PROTEÍNA'].values
                self.nota_nutricional = df['NOTA_NUTRICIONAL'].values
                self.fibras = df['FIBRA ALIMENTAR'].values

                self.limite_calorias_min = limite_calorias_min
                self.limite_calorias_max = limite_calorias_max

            def _evaluate(self, x, out, *args, **kwargs):
                total_calorias = np.sum(x * self.calorias, axis=1)
                total_proteina = np.sum(x * self.proteina, axis=1)
                total_fibras = np.sum(x * self.fibras, axis=1)
                total_nota_nutricional = np.sum(x * self.nota_nutricional, axis=1)

                # Objetivos
                out["F"] = np.column_stack([
                    -total_proteina,      # Maximizar proteína
                    total_calorias,       # Controlar calorias
                    -total_fibras,        # Maximizar fibras
                    -total_nota_nutricional  # Maximizar nota nutricional
                ])

                # Restrições
                g1 = total_calorias - self.limite_calorias_min
                g2 = self.limite_calorias_max - total_calorias

                out["G"] = np.column_stack([g1, g2])

        # Configurar problema
        problema = ProblemaAlimentosAvancado(
            df,
            limite_calorias_min,
            limite_calorias_max
        )

        # Configurar algoritmo
        algoritmo = NSGA2(
          pop_size=pop_size,
          sampling=IntegerRandomSampling(),  # Mudança aqui
          crossover=NutricionalCrossover(),
          mutation=NutricionalMutacao(),
          eliminate_duplicates=True
        )

        # Executar otimização
        res = minimize(
            problema,
            algoritmo,
            ('n_gen', 50),
            seed=42,
            verbose=False
        )

        # Processar resultado
        if res.X is None or len(res.X) == 0:
            return selecao_fallback()

        # Usar função de seleção de melhor solução
        alimentos_selecionados = selecionar_melhor_solucao(
            res,
            df,
            limite_calorias_min,
            limite_calorias_max,
            alimentos_por_refeicao
        )

        # Se não encontrar alimentos, usar fallback
        if not alimentos_selecionados:
            return selecao_fallback()

        return alimentos_selecionados

    except Exception as e:
        print(f"Erro no NSGA-II: {e}")
        return selecao_fallback()
```

```python
def criar_nota_nutricional(df, pesos=None):
    # Pesos padrão
    default_pesos = {
        'PROTEÍNA': 0.4,
        'FIBRA ALIMENTAR': 0.3,
        'ENERGIA KCAL': 0.3
    }

    # Permite sobrescrever pesos
    if pesos:
        default_pesos.update(pesos)

    df['NOTA_NUTRICIONAL'] = (
        normalizar_coluna(df['PROTEÍNA']) * default_pesos['PROTEÍNA'] +
        normalizar_coluna(df['FIBRA ALIMENTAR']) * default_pesos['FIBRA ALIMENTAR'] +
        (1 - normalizar_coluna(df['ENERGIA KCAL'])) * default_pesos['ENERGIA KCAL']
    )
    return df
```

```python
def configurar_nsga2(
    df_alimentos,
    paciente,
    tipos_refeicao,
    dias_semana,
    alimentos_utilizados=None,
    pop_size=100,
    n_gen=50
):

    try:
        # Preparar conjunto de alimentos utilizados
        alimentos_utilizados = set(alimentos_utilizados or [])

        # Definir problema de otimização
        problema = ProblemaNutricional(
            df_alimentos,
            paciente,
            tipos_refeicao,
            dias_semana,
            alimentos_utilizados
        )

        # Personalizar sampling para respeitar estrutura do problema
        class NutricionalSampling(Sampling):
            def _do(self, problem, n_samples, **kwargs):
                # Gerar população inicial respeitando restrições
                X = np.full((n_samples, problem.n_var), -1)

                for i in range(n_samples):
                    # Amostrar alimentos para cada dia, tipo de refeição e refeição
                    for dia in range(len(dias_semana)):
                        for tipo_ref in range(len(tipos_refeicao)):
                            for refeicao in range(problema.n_refeicoes):
                                # Filtrar alimentos disponíveis
                                df_filtrado = df_alimentos[
                                    ~df_alimentos['ALIMENTO'].isin(alimentos_utilizados)
                                ]

                                # Filtrar por tipo de refeição
                                df_tipo_refeicao = df_filtrado[
                                    df_filtrado['TIPO_REFEICAO_MAPEADO'] == tipos_refeicao[tipo_ref]
                                ]

                                # Selecionar aleatoriamente
                                if not df_tipo_refeicao.empty:
                                    indice = np.random.choice(df_tipo_refeicao.index)
                                    X[i, (dia * len(tipos_refeicao) * problema.n_refeicoes) +
                                      (tipo_ref * problema.n_refeicoes) +
                                      refeicao] = indice

                return X

        # Configurar crossover personalizado
        class NutricionalCrossover(Crossover):
            def __init__(self):
                super().__init__(2, 2)

            def _do(self, problem, X, **kwargs):
                # Crossover considerando a estrutura do problema
                n_parents, n_matings, n_var = X.shape[0], X.shape[1], X.shape[2]

                # Criar offspring
                offspring = np.full((n_matings, 2, n_var), -1)

                for i in range(n_matings):
                    p1, p2 = X[i, 0], X[i, 1]

                    # Crossover em pontos específicos
                    crossover_points = np.random.choice(
                        range(n_var),
                        size=problema.n_dias,
                        replace=False
                    )

                    offspring[i, 0], offspring[i, 1] = p1.copy(), p2.copy()

                    for point in crossover_points:
                        offspring[i, 0, point], offspring[i, 1, point] = p2[point], p1[point]

                return offspring

        # Configurar mutação personalizada
        class NutricionalMutacao(Mutation):
            def _do(self, problem, X, **kwargs):
                # Taxa de mutação
                prob_mutacao = 0.2

                for i in range(X.shape[0]):
                    # Mutação em locais específicos
                    if np.random.random() < prob_mutacao:
                        # Selecionar índice para mutação
                        indice_mutacao = np.random.randint(0, X.shape[1])

                        # Filtrar alimentos disponíveis
                        df_filtrado = df_alimentos[
                            ~df_alimentos['ALIMENTO'].isin(alimentos_utilizados)
                        ]

                        # Determinar tipo de refeição
                        tipo_ref_idx = (indice_mutacao // problema.n_refeicoes) % len(tipos_refeicao)

                        # Filtrar por tipo de refeição
                        df_tipo_refeicao = df_filtrado[
                            df_filtrado['TIPO_REFEICAO_MAPEADO'] == tipos_refeicao[tipo_ref_idx]
                        ]

                        # Realizar mutação
                        if not df_tipo_refeicao.empty:
                            X[i, indice_mutacao] = np.random.choice(df_tipo_refeicao.index)

                return X

        # Configurar algoritmo com operadores personalizados
        algoritmo = NSGA2(
            pop_size=pop_size,
            sampling=NutricionalSampling(),
            crossover=NutricionalCrossover(),
            mutation=NutricionalMutacao(),
            eliminate_duplicates=True
        )

        # Executar otimização
        resultado = minimize(
            problema,
            algoritmo,
            termination=('n_gen', n_gen),
            seed=42,
            verbose=True
        )

        # Processar resultados
        return processar_resultado_nsga2(resultado, df_alimentos)

    except Exception as e:
        logger.error(f"Erro na otimização NSGA-II: {e}")
        return None
```

```python
def extrair_primeiro_nome(alimento):
    return alimento.split(' ')[0].strip()

def selecionar_alimentos_sem_duplicacao(
    df,
    limite_calorias_min,
    limite_calorias_max,
    alimentos_por_refeicao,
    limite_calorias_min_refeicao=50,  # Novo parâmetro
    alimentos_utilizados=None,
    verbose=False
):
    # Certificar-se de que alimentos_utilizados seja um conjunto
    if alimentos_utilizados is None:
        alimentos_utilizados = set()

    # Filtrar alimentos disponíveis (não utilizados)
    df_disponivel = df[~df['ALIMENTO'].isin(alimentos_utilizados)].copy()

    # Verificar disponibilidade de alimentos
    if df_disponivel.empty:
        print(f"Nenhum alimento disponível após filtro.")
        return []

    # Função auxiliar de normalização
    def normalizar_coluna(serie):
        min_val = serie.min()
        max_val = serie.max()

        # Evitar divisão por zero
        if min_val == max_val:
            return pd.Series(1, index=serie.index)

        return (serie - min_val) / (max_val - min_val)

    # Estratégias de seleção com pesos mais detalhados
    def calcular_nota_nutricional(alimento):
        # Pesos mais granulares
        pesos_tipo = {
            'FRUTAS': 0.4,
            'LEITE E DERIVADOS': 0.5,
            'CEREAIS, PÃES, TUBÉRCULOS E RAÍZES': 0.3,
            'CARNES E OVOS': 0.6,
            'PEIXES E FRUTOS DO MAR': 0.6,
            'HORTALIÇAS (LEGUMES E VERDURAS)': 0.4,
            'LEGUMINOSAS': 0.5,
            'OUTROS': 0.2,
            'AÇÚCARES E DOCES': 0.1
        }

        # Peso base do tipo
        peso_tipo = pesos_tipo.get(alimento['TIPO'], 0.2)

        # Normalizar calorias
        nota_calorica = 1 - normalizar_coluna(
            pd.Series([alimento['ENERGIA KCAL']])
        )[0]

        # Combinar nota calórica com peso do tipo
        return (nota_calorica * 0.5) + (peso_tipo * 0.5)

    # Estratégias de seleção
    estrategias = [
        # Estratégia 1: Filtro calórico específico para refeição
        lambda df: df[
            (df['ENERGIA KCAL'] >= limite_calorias_min_refeicao / alimentos_por_refeicao) &
            (df['ENERGIA KCAL'] <= limite_calorias_max_refeicao / alimentos_por_refeicao)
        ],

        # Estratégia 2: Ampliar margem calórica
        lambda df: df[
            (df['ENERGIA KCAL'] >= limite_calorias_min_refeicao / alimentos_por_refeicao * 0.8) &
            (df['ENERGIA KCAL'] <= limite_calorias_max_refeicao / alimentos_por_refeicao * 1.2)
        ],

        # Estratégia 3: Todos os alimentos disponíveis
        lambda df: df
    ]

    # Executar seleção de alimentos
    try:
        # Tentar cada estratégia
        for estrategia in estrategias:
            # Criar uma cópia para evitar SettingWithCopyWarning
            df_disponivel_filtrado = estrategia(df_disponivel).copy()

            # Debug: Imprimir detalhes dos alimentos disponíveis
            if verbose:
                print(f"\nAlimentos disponíveis com estratégia: {estrategia.__name__}")
                print(df_disponivel_filtrado[['ALIMENTO', 'TIPO', 'ENERGIA KCAL']].head())
                print(f"Total de alimentos: {len(df_disponivel_filtrado)}")

            # Pular se não houver alimentos
            if df_disponivel_filtrado.empty:
                continue

            # Calcular nota nutricional para cada alimento
            df_disponivel_filtrado.loc[:, 'NOTA_NUTRICIONAL'] = df_disponivel_filtrado.apply(
                calcular_nota_nutricional, axis=1
            )

            # Ordenar por nota nutricional
            df_ordenado = df_disponivel_filtrado.sort_values('NOTA_NUTRICIONAL', ascending=False)

            # Evitar duplicação de primeiro nome e tipo
            alimentos_selecionados = []
            nomes_usados = set()
            tipos_usados = set()
            calorias_total = 0

            for _, alimento in df_ordenado.iterrows():
                primeiro_nome = extrair_primeiro_nome(alimento['ALIMENTO'])
                tipo = alimento['TIPO']
                calorias = alimento['ENERGIA KCAL']

                # Condições mais flexíveis para seleção
                condicoes = [
                    primeiro_nome not in nomes_usados,
                    tipo not in tipos_usados,
                    # Evitar repetição de tipos próximos
                    not any(
                        tipo.startswith(t) or t.startswith(tipo)
                        for t in tipos_usados
                    ),
                    # Verificar limite de calorias
                    calorias_total + calorias <= limite_calorias_max_refeicao
                ]

                if all(condicoes):
                    alimentos_selecionados.append({
                        'ALIMENTO': alimento['ALIMENTO'],
                        'Quantidade': 1,
                        'ENERGIA KCAL': calorias,
                        'Proteína': alimento.get('PROTEÍNA', 0),
                        'Tipo': tipo
                    })
                    nomes_usados.add(primeiro_nome)
                    tipos_usados.add(tipo)
                    calorias_total += calorias

                if len(alimentos_selecionados) == alimentos_por_refeicao:
                    break

            # Verificar se atingiu o mínimo de calorias
            if (
                len(alimentos_selecionados) > 0 and
                calorias_total >= limite_calorias_min_refeicao
            ):
                # Debug: Imprimir alimentos selecionados
                if verbose:
                    print("\nAlimentos Selecionados:")
                    for alimento in alimentos_selecionados:
                        print(f"- {alimento['ALIMENTO']} (Tipo: {alimento['Tipo']}, Calorias: {alimento['ENERGIA KCAL']})")
                    print(f"Total de Calorias: {calorias_total}")

                return alimentos_selecionados

        # Se nenhuma estratégia funcionou
        print("Nenhuma estratégia de seleção de alimentos funcionou.")
        return []

    except Exception as e:
        print(f"Erro na seleção de alimentos: {e}")
        return []
```

Loop Principal do NSGA2

```python
# Dicionário para armazenar refeições de cada dia
refeicoes_semana = {}

# Dicionário para armazenar calorias por dia
calorias_por_dia = {}

# Conjunto global para controlar alimentos utilizados na semana
alimentos_utilizados_semana = set()

for dia in dias_semana:
    # Reiniciar conjunto de alimentos utilizados no dia
    alimentos_utilizados_dia = set()

    # Lista para armazenar refeições do dia
    refeicoes_dia = []

    # Variável para rastrear total de calorias do dia
    total_calorias_dia = 0

    for ref in Tipos_refeicao[:paciente.getConsistencia_alimentar_aceita()]:
        # Criar uma nova instância de Refeição
        refeicao = Refeicao(ref, limite_calorias_max_refeicao, alimentos_por_refeicao, dia)

        # Combinar alimentos utilizados no dia e na semana
        alimentos_utilizados_combinados = (
            alimentos_utilizados_dia.union(alimentos_utilizados_semana)
        )

        # Filtrar alimentos disponíveis (não utilizados)
        df_disponivel = df[df['TIPO_REFEICAO_MAPEADO'] == ref]
        df_disponivel = df_disponivel[
            ~df_disponivel['ALIMENTO'].isin(alimentos_utilizados_combinados)
        ]

        # Verificar disponibilidade de alimentos
        if df_disponivel.empty:
            print(f"Nenhum alimento disponível para a refeição: {ref}")
            continue

        # Selecionar alimentos sem duplicação
        alimentos_selecionados = selecionar_alimentos_sem_duplicacao(
            df_disponivel,
            limite_calorias_min_refeicao,
            limite_calorias_max_refeicao,
            alimentos_por_refeicao,
            alimentos_utilizados=alimentos_utilizados_combinados,
            verbose=False
        )
        print(alimentos_selecionados)
        # Adicionar alimentos à refeição
        alimentos_adicionados = 0
        for alimento in alimentos_selecionados:
            if refeicao.adicionar_alimento2(alimento):
                # Adicionar aos conjuntos de alimentos utilizados
                alimentos_utilizados_dia.add(alimento['ALIMENTO'])
                alimentos_utilizados_semana.add(alimento['ALIMENTO'])

                # Incrementar contadores
                alimentos_adicionados += 1
                total_calorias_dia += alimento['ENERGIA KCAL']

            # Parar se atingir o número desejado de alimentos
            if alimentos_adicionados == alimentos_por_refeicao:
                break

        # Adicionar refeição apenas se tiver alimentos
        if alimentos_adicionados > 0:
            refeicoes_dia.append(refeicao)
        else:
            print(f"Não foi possível adicionar alimentos para a refeição: {ref}")

    # Armazenar refeições do dia
    refeicoes_semana[dia] = refeicoes_dia

    # Registrar informações de calorias do dia
    calorias_por_dia[dia] = {
        'total': total_calorias_dia,
        'meta': paciente.getTMB(),
        'refeicoes': refeicoes_dia
    }
```

```python
def gerar_tabelas_latex_dieta_semanal(refeicoes_semana, calorias_por_dia):
    """
    Gera tabelas LaTeX para cada dia da dieta semanal

    Args:
        refeicoes_semana (dict): Dicionário com refeições de cada dia
        calorias_por_dia (dict): Dicionário com informações de calorias por dia

    Returns:
        str: Código LaTeX com tabelas dos dias
    """
    # String para armazenar o código LaTeX
    latex_tabelas = ""

    # Processar cada dia da semana
    for dia, refeicoes in refeicoes_semana.items():
        # Início da tabela para o dia
        latex_tabelas += fr"""
\begin{{table}}[h]
\caption{{{dia}}}
\centering
\begin{{tabular}}{{llrr}}
\toprule
Refeição & Alimento & Quantidade & Calorias (kcal) \\
\midrule
"""

        # Adicionar linhas de alimentos para cada refeição
        for refeicao in refeicoes:
            for alimento in refeicao.alimentos:
                latex_tabelas += (
                    f"{refeicao.nome} & "
                    f"{obter_nome_alimento(alimento)} & "
                    f"{alimento['Quantidade']:.0f} & "
                    f"{alimento['ENERGIA KCAL']:.0f} \\\\\n"
                )

        # Linha de total de calorias do dia
        latex_tabelas += fr"""
\midrule
\multicolumn{{3}}{{l}}{{Total de Calorias do Dia}} & {calorias_por_dia[dia]['total']:.0f} \\
\bottomrule
\end{{tabular}}
\end{{table}}

"""

    return latex_tabelas

# Versão alternativa com mais opções
def gerar_tabelas_latex_dieta_semanal_v2(refeicoes_semana, calorias_por_dia, opcoes=None):
    """
    Gera tabelas LaTeX para cada dia da dieta semanal com opções personalizáveis

    Args:
        refeicoes_semana (dict): Dicionário com refeições de cada dia
        calorias_por_dia (dict): Dicionário com informações de calorias por dia
        opcoes (dict): Opções de formatação

    Returns:
        str: Código LaTeX com tabelas dos dias
    """
    # Configurações padrão
    configuracoes = {
        'formato_tabela': 'tabular',  # Pode ser 'tabular' ou 'longtable'
        'incluir_meta_calorica': True,
        'estilo_linhas': False,
        'ajustar_largura': False
    }

    # Atualizar configurações com opções fornecidas
    if opcoes:
        configuracoes.update(opcoes)

    # String para armazenar o código LaTeX
    latex_tabelas = ""

    # Processar cada dia da semana
    for dia, refeicoes in refeicoes_semana.items():
        # Início da tabela conforme formato
        if configuracoes['formato_tabela'] == 'longtable':
            latex_tabelas += fr"""
\begin{{longtable}}{{llrr}}
\caption{{{dia}}} \\
\toprule
Refeição & Alimento & Quantidade & Calorias (kcal) \\
\midrule
\endfirsthead
\toprule
Refeição & Alimento & Quantidade & Calorias (kcal) \\
\midrule
\endhead
"""
        else:
            latex_tabelas += fr"""
\begin{{table}}[H]
\caption{{{dia}}}
\centering
\begin{{tabular}}{{llrr}}
\toprule
Refeição & Alimento & Quantidade & Calorias (kcal) \\
\midrule
"""

        # Adicionar linhas de alimentos para cada refeição
        for refeicao in refeicoes:
            for alimento in refeicao.alimentos:
                # Adicionar estilo de linha se configurado
                if configuracoes['estilo_linhas']:
                    latex_tabelas += r"\rowcolor{gray!10} "

                latex_tabelas += (
                    f"{refeicao.nome} & "
                    f"{obter_nome_alimento(alimento)} & "
                    f"{alimento['Quantidade']:.0f} & "
                    f"{alimento['Calorias']:.0f} \\\\\n"
                )

        # Linha de total de calorias do dia
        latex_tabelas += fr"""
\midrule
\multicolumn{{3}}{{l}}{{Total de Calorias do Dia}} & {calorias_por_dia[dia]['total']:.0f} \\
"""

        # Incluir meta calórica se configurado
        if configuracoes['incluir_meta_calorica']:
            latex_tabelas += fr"""
\multicolumn{{3}}{{l}}{{Meta Calórica}} & {calorias_por_dia[dia]['meta']:.0f} \\
"""

        # Fechar tabela conforme formato
        if configuracoes['formato_tabela'] == 'longtable':
            latex_tabelas += r"""\bottomrule
\end{longtable}

"""
        else:
            latex_tabelas += r"""\bottomrule
\end{tabular}
\end{table}

"""

    return latex_tabelas

# Funções de uso
def obter_nome_alimento(alimento):
    """
    Extrai um nome formatado para o alimento

    Args:
        alimento (dict): Dicionário com informações do alimento

    Returns:
        str: Nome formatado do alimento
    """
    try:
        # Split do nome pelo vírgula
        partes = alimento['ALIMENTO'].split(',')

        # Captura primeira palavra de cada parte
        nome_primeira_parte = partes[0].strip()

        # Tratar caso haja mais de uma parte
        if len(partes) > 1:
            nome_segunda_parte = partes[1].strip().split()[0].capitalize()
            nome_alimento = f"{nome_primeira_parte} {nome_segunda_parte}"
        else:
            nome_alimento = nome_primeira_parte

        # Remove parênteses e 'C/'
        nome_alimento = nome_alimento.replace('(', '').replace('C/', '')
        nome_alimento = nome_alimento.upper()
        return nome_alimento.strip()

    except (IndexError, KeyError):
        # Tratamento de erro caso a estrutura do alimento seja diferente
        return "Alimento Desconhecido"

# Uso das funções
# Versão básica
tabelas_latex = gerar_tabelas_latex_dieta_semanal(refeicoes_semana, calorias_por_dia)
print(tabelas_latex)

#
```
